```python
#histogram

plt.hist(df['PM2.5'], bins=10, edgecolor='black')
plt.xlabel('PM2.5 Level')
plt.ylabel('Frequency')
plt.title('Distribution of PM2.5 Levels')
plt.show()
```

**CONCLUSION:**
 The histogram shows the distribution of PM2.5 levels in the dataset. It is indicating the frequency of different PM2.5 concentration ranges. This helps in understanding the typical range of PM2.5 levels observed in the data.

```python
#box plot

import seaborn as sns
sns.boxplot(x=df['Hour'], y=df['PM2.5'] ,hue=df['Hour']<=18)
plt.xlabel('Hour of the Day')
plt.ylabel('PM2.5 Level')
plt.title('Day vs. Night PM2.5 Levels')
plt.show()
```

**CONCLUSION:**
PM2.5 refers to tiny particles in the air which are smaller than 2.5 micrometers.They come from sources like vehicle emissions and wildfires. The graph shows the comparison of PM2.5 levels during day and night.It shows that air quality changes a lot throughout the day as compared to the night.


```python
df['Timestamp'] = pd.to_datetime(df['Timestamp'])
df['Weekday'] = df['Timestamp'].dt.dayofweek < 5
average_pm25 = df.groupby(['Timestamp', 'Weekday'])['PM2.5'].mean().reset_index()
sns.lineplot(x='Timestamp', y='PM2.5', hue='Weekday', data=average_pm25)
plt.xlabel('Timestamp')
plt.ylabel('Average PM2.5 Level')
plt.title('Average PM2.5 Levels between Weekdays and Weekends Over Time')
plt.xticks(rotation=45)
plt.show()
```

**CONCLUSION:**
This graph shows the comaprison of average of PM2.5 levels on weekends and weekdays.We can see that pollution is usually higher on weekdays because more people are driving and working and on weekends, when people do fewer activities that produce pollution, the air quality is usually better. This tells us that human actions affect the nature to a graet extend.

```python
#Heatstamp

heatmap_data = df.pivot_table(index='Hour', columns='Day', values='PM2.5', aggfunc=np.mean)
sns.heatmap(heatmap_data, cmap='YlGnBu')
plt.xlabel('Day')
plt.ylabel('Hour')
plt.title('PM2.5 Levels by Hour and Day')
plt.show()
```

**CONCLUSION:**
This heatmap shows the variation in PM2.5 levels across different hours of the day and days of the month,It helps us to identify any patterns or trends in air quality based on time of day and day of the month.

```python
#Bar plat
sns.barplot(x='Month', y='PM2.5', data=df)
plt.xlabel('Month')
plt.ylabel('Average PM2.5 Level')
plt.title('Average PM2.5 Levels by Month')
plt.show()
```

**CONCLUSION:**
This bar plot shows the relation between average of PM2.5 levels and months. We can see how air pollution changes throughout the year. Some months have higher pollution levels compared to others. This can be because of different factors like weather or human activities.
